# 介紹

字典，在python中用於查表。  
- 字典的前後是用大括號{}包住  
- 裡面每個元素用逗點(,)隔開
- 每個元素都是key:value的形式

```python
students={'Peter':175,'Mary':168,'Keven':165}
#查學生表格，使用關鍵字(key)Keven來查
print(students['Keven'])
```

```python
students={
    'Peter':[175,70,'高雄'],
    'Mary':[168,50,'嘉義'],
    'Keven':[165,55,'高雄']
}
print (students['Mary'])
print (students['Mary'][0]) #取得身高
print (students['Mary'][2]) #取得戶籍
```

```python
print('Peter' in students)
print('peter' in students)
print('Pete' in students)
```

```python
#新增John到students中
students['John']=[180,99,'台北']
print(students)
```

```python
#由students中移除Keven
del students['Keven']
print(students)
```
