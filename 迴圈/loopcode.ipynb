# Loop

Loop 指令可以讓程式碼依我的條件不斷執行，直到條件滿足為止。

```python
# 底下的程式碼會印出0-9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
for i in range(10):
    print(i+1)
```

# Example

計算 1+2+3+...+10

```python
sum=0
for i in range(10):
    sum=sum+(i+1)
print(sum)    
```

印出底下圖形

&  
&&  
&&&  
&&&&  
&&&&&  

```python
# print('&'*1)
# print('&'*2)
# print('&'*3)
# print('&'*4)
# print('&'*5)

for i in range(6):
    print("&"*(i+1))
```

印出
1\*1=1  
1\*2=2  
1\*3=3  
1\*4=4  
...  
1\*9=9

```python
# print(f'1*1={1*1}')
# print(f'1*2={1*2}')
# print(f'1*3={1*3}')
# print(f'1*4={1*4}')

for i in range(10):
    print(f'1*{(i+1)}={1*(i+1)}')
```

# Range

python:
- 懶惰是一種美德
- 由0開始
- 包含頭，不包含尾

```python
print(list(range(5)))
#區間，由0開始，到5-1結束
print(list(range(0,5)))

#區間，由1開始，到5-1結束
print(list(range(1,5)))

#區間，由1開始，到5-1結束，每次跳兩個
print(list(range(1,5,2)))
```

```python
print(list(range(2,101,2)))
```
